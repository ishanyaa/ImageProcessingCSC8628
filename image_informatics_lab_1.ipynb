## numpy

### The Basics

```
a = np.array([2,3,4])
print(a)
print(a.dtype)
print(a.shape)
print(a.size)
```

### Array Creation

```
a = np.zeros((3, 4))
print(a)
print(a.shape)

a = np.ones((3, 4))
print(a)
```

```
c = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print(c)
print(c.size)
print(c.shape)
```

```
a = np.ones((2,3,4), dtype=np.int16)
print(a)
print(a.dtype)
print(a.size)
print(a.shape)
```

### Basic Operations

```
a = np.array([20,30,40,50])
b = np.arange(4)
print(a)
print(b)

c = a-b
print(c)

d = a<35
print(d)
```

```
a = np.array([[1,1], [0,1]])
print(a)
print(a.size)
print(a.shape)

b = np.array([[2,0], [3,4]])
print(b)
print(b.size)
print(b.shape)

c = a*b # elementwise product

print(c)
```

### Universal Functions

```
a = np.arange(3)
print(a)

b = np.exp(a)
print(b)

c = np.sqrt(a)
print(c)
```

### Indexing, Slicing and Iterating

```
a = np.arange(10)**3

print(a.shape)
print(a.size)
print(a)
print(a[2])
print(a[2:5])


# a[start:stop:step] -- replaced with 1000
a[:6:2] = 1000
print(a)

# reverse array
print(a[ : :-1])

for i in a:
  print(i**(1/3.))


# to round of to x decimal values

for i in a:
  print(round(i**(1/3.), 2))
```

```
a = np.random.rand(5,4)
print(a)
print(a[2,2])

print(a[0:2,2])
print(a[:,2])

print(a[1:3,:])
print(a[-1])
```

### Indexing with Boolean Arrays

```
a = np.arange(12).reshape(3,4)
b = a > 4
print(b)

c = a[b]
print(c)
```

### Multiple figures and axes

```
def f(t):
  return np.exp(-t) * np.cos(2*np.pi*t)
  #return np.exp(-t) # exp function
  #return np.cos(2*np.pi*t) #cos function


t1 = np.arange(0.0, 5.0, 0.1)
t2 = np.arange(0.0, 5.0, 0.02)

plt.figure()
plt.subplot(211)
plt.plot(t1, f(t1), 'bo', t2, f(t2), 'k')

plt.subplot(212)
plt.plot(t2, np.cos(2*np.pi*t2), 'r--')
plt.show()
```

### Working with text

```
mu, sigma = 100, 15
x = mu + sigma * np.random.randn(10000)

# the histogram of the data
n, bins, patches = plt.hist(x, 50, density=1, facecolor='g', alpha=0.75)

print(len(n))
print(len(bins))
print(len(patches))

plt.xlabel('Smarts')
plt.ylabel('Probability')
plt.title('Histogram of IQ')

plt.text(60, .025, r'$\mu=100,\ \sigma=15$')
plt.axis([40, 160, 0, 0.03])
plt.grid(True)
plt.show()
```

## scikit-image

### Basics

```
from skimage import data

camera = data.camera()
p = plt.imshow(camera)
print(camera.shape)
print(camera.min())
print(camera.max())
print(camera.mean())
print(camera[10, 20])
```

```
mask = camera < 87
camera[mask] = 255
p = plt.imshow(camera)
```
